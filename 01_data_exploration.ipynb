```python
import pandas as pd
import ast
from collections import Counter

# Load raw datasets
df_postings = pd.read_csv('job_postings.csv')
df_clean = pd.read_csv('final_clean_jobs_dataset.csv')
df_ai = pd.read_csv('AI_Job_Market_Trends_2026.csv')

print(" RAW DATA LOADED ")
print(f"Postings shape: {df_postings.shape}")
print(f"Clean jobs shape: {df_clean.shape}")
print(f"AI trends shape: {df_ai.shape}")
```

```python
# Clean job postings
df_postings = df_postings.drop_duplicates()
df_postings['job_location'] = df_postings['job_location'].fillna('Unknown').str.strip().str.title()
df_postings['job_level'] = df_postings['job_level'].fillna('Not Specified').str.strip()
if 'job_title' in df_postings.columns:
    df_postings['job_title'] = df_postings['job_title'].str.strip().str.title()

# Clean details dataset
df_clean = df_clean.drop_duplicates()
if 'skills' in df_clean.columns:
    df_clean['skills'] = df_clean['skills'].fillna('[]')

# Clean AI trends dataset
df_ai = df_ai.drop_duplicates()
if 'remote_type' in df_ai.columns:
    df_ai['remote_type'] = df_ai['remote_type'].fillna('Unknown').str.strip().str.capitalize()

print(" DATA CLEANING COMPLETE ")
```

```python
import pandas as pd
import ast
from collections import Counter

# Load your datasets
df_postings = pd.read_csv('job_postings.csv')
df_clean = pd.read_csv('final_clean_jobs_dataset.csv')
df_ai = pd.read_csv('AI_Job_Market_Trends_2026.csv')


print("JOB MARKET ANALYTICS PROJECT — MULTI-QUESTION INSIGHTS")


# QUESTION 1: Which programming languages and tools pop up the most?
skill_cols = [col for col in df_ai.columns if col.startswith('skills_')]
ai_skill_counts = df_ai[skill_cols].sum().reset_index()
ai_skill_counts.columns = ['skill', 'count']
ai_skill_counts['skill'] = ai_skill_counts['skill'].str.replace('skills_', '').str.capitalize()
ai_skill_counts = ai_skill_counts.sort_values(by='count', ascending=False)

print("\n[Q1] Top In-Demand Skills & Tools (AI Trends Dataset):")
print(ai_skill_counts.to_string(index=False))


# QUESTION 2: How many years of experience / seniority levels do employers ask for?
print("\n[Q2] Job Seniority / Levels Breakdown (`job_postings`):")
if 'job_level' in df_postings.columns:
    print(df_postings['job_level'].value_counts().head(10))
else:
    print("Column 'job_level' not found in df_postings.")


#  QUESTION 3: Where are most of these jobs located?
print("\n[Q3] Top Job Locations (`job_postings`):")
if 'job_location' in df_postings.columns:
    print(df_postings['job_location'].value_counts().head(10))
else:
    print("Column 'job_location' not found in df_postings.")


#  QUESTION 4: Skill Clustering & Tech Stacks (Co-occurrence)
print("\n[Q4] Python & SQL Co-occurrence (AI Trends Dataset):")
print(pd.crosstab(df_ai['skills_python'], df_ai['skills_sql'],
                  rownames=['Python Required'], colnames=['SQL Required']))


#  QUESTION 5: Workplace Flexibility vs. Tech Stack ---
print("\n[Q5] Remote Type Distribution vs. Python Requirement:")
print(pd.crosstab(df_ai['remote_type'], df_ai['skills_python'],
                  rownames=['Remote Type'], colnames=['Python Required (0=No, 1=Yes)']))

```

```python
df_postings.to_csv('cleaned_job_postings.csv', index=False)
df_clean.to_csv('cleaned_jobs_detail.csv', index=False)
df_ai.to_csv('cleaned_ai_job_trends.csv', index=False)

print(" CLEANED FILES EXPORTED SUCCESSFULLY ")
```
